# 🟢 Easy: Implement Softmax

Implement the **Softmax** function from scratch.

$$\text{softmax}(x_i) = \frac{e^{x_i}}{\sum_j e^{x_j}}$$

### Signature
```python
def my_softmax(x: torch.Tensor, dim: int = -1) -> torch.Tensor:
    ...
```

### Rules
- Do **NOT** use `torch.softmax`, `F.softmax`, or `torch.nn.Softmax`
- Must be **numerically stable** (hint: subtract `max` before `exp`)

### Example
```
Input:  tensor([1., 2., 3.])
Output: tensor([0.0900, 0.2447, 0.6652])  # sums to 1.0
```

### 補足：`x - x.max()` (数値安定性)
`x`に大きな数値(例えば1000)があると、`torch.exp(1000)`が`inf`に発散してしまう。最大値を引けば最大要素が0になるので`exp`の出力の上限が1に抑えられ、オーバーフローを抑えることが出来る。分子と分母に定数倍がかかるだけなので、数学的には等価となる。softmax()を書く時の定石。

````python
import torch
````

````python
# ✏️ YOUR IMPLEMENTATION HERE

def my_softmax(x: torch.Tensor, dim: int = -1) -> torch.Tensor:
    x_shifted = x - x.max(dim=dim, keepdim=True).values
    e_x = torch.exp(x_shifted)
    return e_x / e_x.sum(dim=dim, keepdim=True)
````

````python
# 🧪 Debug
x = torch.tensor([1.0, 2.0, 3.0])
print(torch.exp(x.max()))
print(x.sum())
print("Output:", my_softmax(x, dim=-1))
print("Sum:   ", my_softmax(x, dim=-1).sum())  # should be ~1.0
print("Ref:   ", torch.softmax(x, dim=-1))
````

- `dim`の指定で「どの方向につぶすか」が決まる

````python
print(tmp.sum(dim=0))  # tensor(5., 7., 9.) → 「行」方向につぶす
print(tmp.sum(dim=1))  # tensor(6., 15.) → 「列」方向につぶす
print(tmp.sum(dim=-1)) # tensor(6., 15.) # 最後の次元について縮約する
````

- `keepdim=True`で次元を残せる

````python
print("Tensor of False:", tmp.sum(dim=1, keepdim=False))
print("Tensor of True:", tmp.sum(dim=1, keepdim=True))

print("Shape of False:", tmp.sum(dim=1, keepdim=False).shape)
print("Shape of True:", tmp.sum(dim=1, keepdim=True).shape)
````
