# 🟢 Easy: GELU Activation

Implement the **GELU** (Gaussian Error Linear Unit) activation.

$$\text{GELU}(x) = x \cdot \Phi(x) = x \cdot 0.5 \cdot (1 + \text{erf}(x / \sqrt{2}))$$

### Signature
```python
def my_gelu(x: Tensor) -> Tensor: ...
```

### Rules
- Do NOT use `F.gelu`, `nn.GELU`, or `torch.nn.functional.gelu`
- Use `torch.erf` for the exact version

#### GELUとは
活性化関数の一つ。Gaussian Error Linear Unitの略。
ニューラルネットの非線形性を担う。
- 数式:
  $$
  \text{GELU}(x) = x\cdot\Phi(x)
  $$\
  $\Phi(x)$は標準正規分布の累積分布関数(CDF) \
  「入力xに、それが正規分布上で生き残る確率を掛ける」操作。

#### ReLUとの比較
|   |ReLU    |GELU    |
|---|--------|--------|
|式 |max(0, x)|$x\cdot\Phi(x)$|
|形 |0で折れ曲がる|滑らかな曲線 |
|負の入力|完全に0 |0付近は少しの負の値を通す|

#### Deep Learningでの使われ方
Transformer系モデルの標準的な活性化関数。BERT、GPT-2、ViTなどのFeed-Forward層(MLPブロック)で採用されている。滑らかで微分が連続なため勾配ベースの最適化と相性が良く、大規模モデルでReLUより良い性能を示すことが多い。

#### 実装での注意
$\Phi(x)$は誤差関数`torch.erf`で計算できる：
$$
\Phi(x)=\frac{1}{2}\left(1 + \operatorname{erf}\left(x / \sqrt{2}\right)\right)
$$

````python
import torch
import math
````

````python
# ✏️ YOUR IMPLEMENTATION HERE

def my_gelu(x):
    return 0.5 * x * (1.0 + torch.erf(x / math.sqrt(2.0)))
````

````python
# 🧪 Debug
x = torch.tensor([-2., -1., 0., 1., 2.])
print('Output:', my_gelu(x))
print('Ref:   ', torch.nn.functional.gelu(x))
````
